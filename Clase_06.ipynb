# Orden de esquemas

O(H) Número fijo de alelos 0 o 1 de un esquema.


# Longitud 

d(H) Distancia entre el primer y el último elemento fijo.

# Función

m(H, t) Cantidad de elementos de un esquema H en la generación t.



# Población de un esquema en la generación t+1

$$ m(H, t+1) = m(H, t) \cdot \frac{f(H)}{\hat{f}} $$
