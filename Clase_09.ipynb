# Selección por torneo
1. Se escogen dos individuos aleatoriamente de la población.
2. Se genera un número aleatorio $r$ en $[0, 1]$
3. Si $r<umbral$, el menor de los dos es seleccionado como padre, de otra forma el peor elemento es seleccionado.

Si se toma el valor de umbral de $0.5$ es la misma probabilidad para los dos. Se podría comenzar con un valor de umbral de $0.7$ o $0.8$ e ir incrementando después de varias generaciones hasta $0.9$ ya que si lo fijamos en $1$, siempre se elegirá al mejor elemento.
