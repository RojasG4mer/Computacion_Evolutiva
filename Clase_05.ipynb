# Teoría de esquemas

Esquema: Patrón de similaridad que describe un conjunto de individuos dentro de la población.

ALFABETO $= \{0, 1, * \}$

Donde:
- $3^l$ esquemas posibles.
- $l$ longitud del cromosoma (número de alelos).
- El * nos dice que alelo no nos interesa lo que valga.

Nos ayuda a ver que tanto un alelo afecta a la población en general.
